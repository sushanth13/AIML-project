```python
# MAMBA IMAGE CAPTIONING

import os
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.transforms as transforms
import timm
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import random
import re
import matplotlib.pyplot as plt
from collections import Counter
from mamba_ssm import Mamba
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)


# Settings

IMAGE_DIR = "/kaggle/input/datasets/banuprasadb/flickr30k/Images"
CAPTION_FILE = "/kaggle/input/datasets/banuprasadb/flickr30k/captions.txt"

MAX_LEN = 40
EMBED_DIM = 768
BATCH_SIZE = 32   
EPOCHS = 30
LR = 1e-4         


# Loading captions

def load_captions(file):

    captions = {}

    with open(file, 'r', encoding='utf-8') as f:

        for i, line in enumerate(f):

            if i == 0:
                continue

            img, caption = line.strip().split(",", 1)

            img_id = img.split("#")[0]

            caption = caption.lower()
            caption = re.sub(r"[^\w\s]", "", caption)

            caption = "<start> " + caption + " <end>"

            captions.setdefault(img_id, []).append(caption)

    return captions

captions = load_captions(CAPTION_FILE)

valid_ids = [k for k in captions if os.path.exists(os.path.join(IMAGE_DIR, k))]
random.shuffle(valid_ids)

split = int(0.8 * len(valid_ids))
train_ids = valid_ids[:split]
test_ids = valid_ids[split:]


# Building vocabulary

all_words = []
for img in valid_ids:
    for cap in captions[img]:
        all_words.extend(cap.split())

counter = Counter(all_words)

vocab = ["<pad>", "<unk>"] + [w for w, _ in counter.most_common(20000)]

word2idx = {w: i for i, w in enumerate(vocab)}
idx2word = {i: w for w, i in word2idx.items()}

vocab_size = len(vocab)

PAD_IDX = word2idx["<pad>"]
UNK_IDX = word2idx["<unk>"]
END_IDX = word2idx["<end>"]


# Image transformation

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# Dataset

class FlickrDataset(Dataset):

    def __init__(self, ids):
        self.ids = ids

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):

        img_id = self.ids[idx]
        img_path = os.path.join(IMAGE_DIR, img_id)

        image = Image.open(img_path).convert("RGB")
        image = transform(image)

        caption = random.choice(captions[img_id])
        tokens = caption.split()

        token_ids = [word2idx.get(t, UNK_IDX) for t in tokens]

        token_ids = token_ids[:MAX_LEN]
        token_ids += [PAD_IDX] * (MAX_LEN - len(token_ids))

        input_seq = torch.tensor(token_ids[:-1])
        target_seq = torch.tensor(token_ids[1:])

        return image, input_seq, target_seq


train_loader = DataLoader(
    FlickrDataset(train_ids),
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)


# Model

class MambaCaptionModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = timm.create_model(
            "efficientnet_b0",
            pretrained=True,
            num_classes=0
        )

        # freeze initially
        for p in self.encoder.parameters():
            p.requires_grad = False

        self.image_proj = nn.Linear(self.encoder.num_features, EMBED_DIM)

        self.embedding = nn.Embedding(vocab_size, EMBED_DIM)

        self.dropout = nn.Dropout(0.1)

        self.mamba1 = Mamba(d_model=EMBED_DIM, d_state=64, d_conv=4, expand=2)
        self.mamba2 = Mamba(d_model=EMBED_DIM, d_state=64, d_conv=4, expand=2)

        self.fc = nn.Linear(EMBED_DIM, vocab_size)

    def forward(self, images, captions):

        img_feat = self.encoder(images)
        img_feat = self.image_proj(img_feat).unsqueeze(1)

        cap_embed = self.embedding(captions)
        cap_embed = self.dropout(cap_embed)

        x = torch.cat([img_feat, cap_embed], dim=1)

        x = self.mamba1(x)
        x = self.mamba2(x)

        x = x[:, 1:, :]

        return self.fc(x)

model = MambaCaptionModel().to(device)


# Calculating loss and optimization

criterion = nn.CrossEntropyLoss(ignore_index=PAD_IDX)

optimizer = optim.AdamW(model.parameters(), lr=LR)


# Accuracy

def token_accuracy(preds, targets):

    preds = preds.argmax(dim=-1)

    mask = targets != PAD_IDX

    correct = (preds == targets) & mask

    return correct.sum().item(), mask.sum().item()


# Training

for epoch in range(EPOCHS):

    # unfreeze earlier
    if epoch == 5:
        print("Unfreezing encoder...")
        for p in model.encoder.parameters():
            p.requires_grad = True

    model.train()

    total_loss = 0
    total_correct = 0
    total_tokens = 0

    for images, inputs, targets in train_loader:

        images = images.to(device)
        inputs = inputs.to(device)
        targets = targets.to(device)

        optimizer.zero_grad()

        outputs = model(images, inputs)

        loss = criterion(
            outputs.reshape(-1, vocab_size),
            targets.reshape(-1)
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

        optimizer.step()

        total_loss += loss.item()

        correct, tokens = token_accuracy(outputs, targets)

        total_correct += correct
        total_tokens += tokens

    print(f"Epoch {epoch+1}/{EPOCHS} | Loss:{total_loss/len(train_loader):.4f} | Acc:{total_correct/total_tokens:.4f}")


# Beam search 

def beam_search_caption(image_path, beam_width=5):

    model.eval()

    image = Image.open(image_path).convert("RGB")
    image_tensor = transform(image).unsqueeze(0).to(device)

    sequences = [[[word2idx["<start>"]], 0.0]]

    for _ in range(MAX_LEN-1):

        all_candidates = []

        for seq, score in sequences:

            input_seq = torch.tensor(seq).unsqueeze(0).to(device)

            outputs = model(image_tensor, input_seq)

            logits = outputs[0, -1]

            logits[UNK_IDX] = -1e9
            logits[PAD_IDX] = -1e9

            # repetition penalty
            for t in set(seq):
                logits[t] /= 1.2

            log_probs = torch.log_softmax(logits, dim=-1)

            top_probs, top_idx = torch.topk(log_probs, beam_width)

            for i in range(beam_width):
                candidate = [seq + [top_idx[i].item()], score + top_probs[i].item()]
                all_candidates.append(candidate)

        sequences = sorted(all_candidates, key=lambda x: x[1]/(len(x[0])**0.7), reverse=True)[:beam_width]

    best_seq = sequences[0][0]

    words = [idx2word[t] for t in best_seq if t not in [word2idx["<start>"], PAD_IDX, END_IDX]]

    return " ".join(words).capitalize() + "."


# Show results

def show_results(n=4):

    samples = random.sample(test_ids, n)

    for img_id in samples:

        path = os.path.join(IMAGE_DIR, img_id)

        pred = beam_search_caption(path)

        img = Image.open(path)

        plt.imshow(img)
        plt.title(pred)
        plt.axis("off")
        plt.show()

        print("TRUE:")
        for c in captions[img_id][:5]:
            print("-", c.replace("<start> ","").replace(" <end>",""))

        print("="*80)


# BLEU

def compute_bleu():

    smoothie = SmoothingFunction().method4

    scores = []

    for img_id in random.sample(test_ids, 200):

        pred = beam_search_caption(os.path.join(IMAGE_DIR, img_id))
        pred_tokens = pred.lower().split()

        refs = [c.replace("<start> ","").replace(" <end>","").split() for c in captions[img_id]]

        score = sentence_bleu(refs, pred_tokens, smoothing_function=smoothie)

        scores.append(score)

    print("BLEU-4:", sum(scores)/len(scores))


# Run and display


show_results()

compute_bleu()
```
