```python
#import kagglehub
import os
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt
import random
import re
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Download Data
print("📥 Downloading Flickr8k...")
#path = kagglehub.dataset_download("adityajn105/flickr8k")
#image_dir = os.path.join(path, "Images")
#caption_file = os.path.join(path, "captions.txt")

image_dir = r"C:\Users\janas\Downloads\flickr8k\Images"
caption_file = r"C:\Users\janas\Downloads\flickr8k\captions.txt"

# 2. Load Captions
def load_captions(filename):
    captions = {}
    with open(filename, "r", encoding="utf-8") as f:
        for line_num, line in enumerate(f):
            line = line.strip()
            if not line or (line_num == 0 and "image" in line.lower()): 
                continue
            try:
                img, caption = line.split(",", 1)
                img_id = img.split("#")[0]
                
                # Clean caption
                caption = re.sub(r"[\"\'`]", "", caption.lower().strip())
                caption = re.sub(r"\s+", " ", caption)
                
                # FIX: Replaced &lt; and &gt; with actual < and >
                caption = f"<start> {caption.strip()} <end>"
                
                captions.setdefault(img_id, []).append(caption)
            except: 
                continue
    return captions

captions = load_captions(caption_file)

# 3. Filter Valid Images
valid_image_ids = []
for img_id in captions:
    img_path = os.path.join(image_dir, img_id)
    if os.path.exists(img_path):
        valid_image_ids.append(img_id)

print(f"Valid images: {len(valid_image_ids)}/{len(captions)} ({len(valid_image_ids)/len(captions)*100:.1f}%)")

# 4. Split Data
random.shuffle(valid_image_ids)
split = int(0.8 * len(valid_image_ids))
train_ids = valid_image_ids[:split]
test_ids = valid_image_ids[split:]

# 5. Tokenizer
# FIX: Filters must allow < and > characters
filters = '!"#$%&()*+,-./:;=?@[\\]^_`{|}~\t\n' 

tokenizer = Tokenizer(num_words=5000, oov_token="<unk>", filters=filters)
all_captions_list = [cap for img_id in valid_image_ids for cap in captions[img_id]]
tokenizer.fit_on_texts(all_captions_list)

MAX_LEN = 35
vocab_size = min(5000, len(tokenizer.word_index) + 1)
index_word = {v: k for k, v in tokenizer.word_index.items()}

# Get token IDs safely
start_token = tokenizer.word_index.get("<start>")
end_token = tokenizer.word_index.get("<end>")

if start_token is None or end_token is None:
    print("⚠️ WARNING: Special tokens not found. Check tokenizer filters.")

print(f" Vocab: {vocab_size}, Max Len: {MAX_LEN}")

# 6. Image Loading Helper
def load_image(path):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, (224, 224))
    img = tf.keras.applications.resnet50.preprocess_input(img)
    return img.numpy()

# 7. Data Generator
def data_generator(image_ids, captions, batch_size=16):
    while True:
        ids_copy = image_ids[:]
        random.shuffle(ids_copy)
        
        batch_img, batch_in, batch_out = [], [], []
        
        for img_id in ids_copy:
            img_path = os.path.join(image_dir, img_id)
            
            try:
                img = load_image(img_path)
                for cap in captions[img_id][:5]:
                    seq = tokenizer.texts_to_sequences([cap])[0]
                    
                    if len(seq) < 2: continue  
                    
                    in_seq = pad_sequences([seq[:-1]], maxlen=MAX_LEN, padding='post')[0]
                    out_seq = pad_sequences([seq[1:]], maxlen=MAX_LEN, padding='post')[0]
                    
                    batch_img.append(img)
                    batch_in.append(in_seq)
                    batch_out.append(out_seq)
                    
                    if len(batch_img) == batch_size:
                        yield (np.array(batch_img), np.array(batch_in)), np.array(batch_out)
                        batch_img, batch_in, batch_out = [], [], []
            except:
                continue
                
        if len(batch_img) > 0: 
            yield (np.array(batch_img), np.array(batch_in)), np.array(batch_out)

# 8. Create Datasets
train_ds = tf.data.Dataset.from_generator(
    lambda: data_generator(train_ids, captions, 16),
    output_signature=(
        (
            tf.TensorSpec(shape=(None, 224, 224, 3), dtype=tf.float32),
            tf.TensorSpec(shape=(None, MAX_LEN), dtype=tf.int32)
        ),
        tf.TensorSpec(shape=(None, MAX_LEN), dtype=tf.int32)
    )
).prefetch(tf.data.AUTOTUNE)

test_ds = tf.data.Dataset.from_generator(
    lambda: data_generator(test_ids, captions, 16),
    output_signature=(
        (
            tf.TensorSpec(shape=(None, 224, 224, 3), dtype=tf.float32),
            tf.TensorSpec(shape=(None, MAX_LEN), dtype=tf.int32)
        ),
        tf.TensorSpec(shape=(None, MAX_LEN), dtype=tf.int32)
    )
).prefetch(tf.data.AUTOTUNE).take(100)

print(" Datasets ready!")

# 9. Model Architecture
def ImageEncoder(embed_dim=256):
    base = tf.keras.applications.ResNet50(include_top=False, weights="imagenet", pooling="avg")
    base.trainable = False
    inputs = layers.Input((224, 224, 3))
    x = base(inputs)
    x = layers.Dense(embed_dim, activation="relu")(x)
    return models.Model(inputs, x)

def create_model(vocab_size, embed_dim=256):
    img_input = layers.Input((224, 224, 3))
    caption_input = layers.Input((MAX_LEN,))
    
    img_features = ImageEncoder(embed_dim)(img_input)
    caption_features = layers.Embedding(vocab_size, embed_dim, mask_zero=True)(caption_input)
    
    img_expanded = layers.RepeatVector(MAX_LEN)(img_features)
    combined = layers.Add()([caption_features, img_expanded])
    
    lstm_out = layers.LSTM(256, return_sequences=True)(combined)
    output = layers.TimeDistributed(layers.Dense(vocab_size))(lstm_out)
    
    return models.Model([img_input, caption_input], output)

model = create_model(vocab_size)
model.summary()

# 10. Loss Function
def masked_loss(y_true, y_pred):
    loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
    loss = loss_fn(y_true, y_pred)
    mask = tf.cast(y_true != 0, tf.float32)
    return tf.reduce_sum(loss * mask) / tf.reduce_sum(mask)

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss=masked_loss,
    metrics=['accuracy']
)

# 11. Train
print("🚀 Training...")
steps_per_epoch = len(train_ids) // 16 
history = model.fit(
    train_ds, 
    epochs=5, 
    steps_per_epoch=steps_per_epoch,
    validation_data=test_ds,
    verbose=1
)

# 12. Generate Captions
def generate_caption(model, image_path, max_length=35):
    img = tf.expand_dims(load_image(image_path), 0)
    caption = [start_token]
    
    for _ in range(max_length):
        seq = pad_sequences([caption], maxlen=MAX_LEN, padding='post')
        pred = model.predict([img, seq], verbose=0)
        next_word_idx = np.argmax(pred[0, len(caption)-1])
        caption.append(next_word_idx)
        if next_word_idx == end_token:
            break
    
    decoded = []
    for w in caption:
        if w in index_word:
            word = index_word[w]
            if word not in ['<start>', '<end>', '<unk>']:
                decoded.append(word)
    return ' '.join(decoded)

# 13. Visualize Results
def show_results(n=6):
    plt.figure(figsize=(15, 10))
    for i in range(n):
        img_id = test_ids[i]
        img_path = os.path.join(image_dir, img_id)
        pred_caption = generate_caption(model, img_path)
        true_captions = [c.replace('<start> ', '').replace(' <end>', '') for c in captions[img_id][:3]]
        
        plt.subplot(n // 2 + (n % 2), 2, i + 1)
        plt.imshow(plt.imread(img_path))
        plt.axis('off')
        text_str = f"**PRED:** {pred_caption}\n\n**TRUE:**\n" + '\n'.join([f'- {c}' for c in true_captions])
        plt.title(text_str, fontsize=10, loc='left')
    
    plt.tight_layout()
    plt.show()

print("🎯 FINAL RESULTS:")
show_results(4)
print("✅ COMPLETE! Model ready.")
```

```python
import os
import random
import math
import matplotlib.pyplot as plt

def show_results(n=6, seed=None):
    """
    Show n random images with predicted and true captions.
    - n: number of images to display (unique, no repeats)
    - seed: set an int for reproducible randomness; leave None for fresh randomness
    """
    if seed is not None:
        random.seed(seed)

    if n > len(test_ids):
        raise ValueError(f"Requested n={n} but only {len(test_ids)} test images are available.")

    # Pick n unique random image IDs
    chosen_ids = random.sample(test_ids, n)

    rows = math.ceil(n / 2)
    cols = 2
    plt.figure(figsize=(15, 5 * rows))

    for i, img_id in enumerate(chosen_ids, start=1):
        img_path = os.path.join(image_dir, img_id)
        pred_caption = generate_caption(model, img_path)

        # Clean up start/end tokens (HTML-escaped in your snippet)
        true_captions = [
            c.replace('&lt;start&gt; ', '').replace(' &lt;end&gt;', '')
            for c in captions[img_id][:3]
        ]

        ax = plt.subplot(rows, cols, i)
        ax.imshow(plt.imread(img_path))
        ax.axis('off')

        # Matplotlib titles don't render Markdown; just make it readable
        title = f"PRED: {pred_caption}\n\nTRUE:\n" + '\n'.join([f"- {c}" for c in true_captions])
        ax.set_title(title, fontsize=10, loc='left')

    plt.tight_layout()
    plt.show()

print("FINAL RESULTS:")
show_results(n=6, seed=None)  # Set seed=42 to reproduce the same random pick
print(" COMPLETE! Model ready.")
```

```python
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

def compute_bleu_scores(reference_captions, predicted_caption):
    """
    reference_captions: list of strings (ground truth)
    predicted_caption: string
    """
    smoothie = SmoothingFunction().method4

    # Tokenize
    refs = [ref.split() for ref in reference_captions]
    hyp = predicted_caption.split()

    bleu1 = sentence_bleu(refs, hyp, weights=(1, 0, 0, 0), smoothing_function=smoothie)
    bleu2 = sentence_bleu(refs, hyp, weights=(0.5, 0.5, 0, 0), smoothing_function=smoothie)
    bleu4 = sentence_bleu(refs, hyp, weights=(0.25, 0.25, 0.25, 0.25), smoothing_function=smoothie)

    return bleu1, bleu2, bleu4
```

```python
import math
import random
import matplotlib.pyplot as plt
import os

def show_results_with_bleu(n=6, seed=None):
    if seed is not None:
        random.seed(seed)

    chosen_ids = random.sample(test_ids, n)

    rows = math.ceil(n / 2)
    cols = 2
    plt.figure(figsize=(15, 5 * rows))

    for i, img_id in enumerate(chosen_ids, start=1):
        img_path = os.path.join(image_dir, img_id)

        # Generate caption
        pred_caption = generate_caption(model, img_path)

        # Ground truth captions
        true_captions = [
            c.replace('<start> ', '').replace(' <end>', '')
            for c in captions[img_id][:3]
        ]

        # BLEU scores
        bleu1, bleu2, bleu4 = compute_bleu_scores(true_captions, pred_caption)

        ax = plt.subplot(rows, cols, i)
        ax.imshow(plt.imread(img_path))
        ax.axis('off')

        title = (
            f"PRED: {pred_caption}\n\n"
            f"TRUE:\n" + '\n'.join([f"- {c}" for c in true_captions]) +
            f"\n\nBLEU-1: {bleu1:.3f} | BLEU-2: {bleu2:.3f} | BLEU-4: {bleu4:.3f}"
        )

        ax.set_title(title, fontsize=10, loc='left')

    plt.tight_layout()
    plt.show()

print("🎯 FINAL RESULTS WITH BLEU SCORES")
show_results_with_bleu(n=6, seed=42)
```
